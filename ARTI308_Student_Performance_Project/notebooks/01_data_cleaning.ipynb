# Cell 1 - Markdown
"""
# 01 Data Cleaning

## Student Performance Prediction Using Machine Learning

This notebook prepares the raw student performance dataset for machine learning.

Main tasks:
- Load the raw dataset
- Inspect dataset structure
- Handle missing values
- Check duplicates
- Correct invalid values
- Create Pass/Fail target
- Validate the cleaned dataset
- Save the cleaned dataset and cleaning summary
"""

```python
# Cell 2 - Import Libraries

import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Cell 3 - Define Paths

PROJECT_ROOT = Path("..")

RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw" / "StudentPerformanceFactors.csv"
PROCESSED_DATA_PATH = PROJECT_ROOT / "data" / "processed" / "StudentPerformanceFactors_cleaned.csv"

TABLES_DIR = PROJECT_ROOT / "outputs" / "tables"
TABLES_DIR.mkdir(parents=True, exist_ok=True)

print("Raw data path:", RAW_DATA_PATH)
print("Processed data path:", PROCESSED_DATA_PATH)
print("Tables directory:", TABLES_DIR)
```

```python
# Cell 4 - Load Dataset

df = pd.read_csv(RAW_DATA_PATH)

df.head()
```

```python
# Cell 5 - Dataset Overview

print("Dataset Shape:")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

print("\nColumn Names:")
for col in df.columns:
    print("-", col)
```

```python
# Cell 6 - Data Information

df.info()
```

```python
# Cell 7 - Check Missing Values

missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage (%)": missing_percentage.round(2)
})

missing_summary[missing_summary["Missing Values"] > 0]
```

```python
# Cell 8 - Handle Missing Values

categorical_missing_columns = [
    "Teacher_Quality",
    "Parental_Education_Level",
    "Distance_from_Home"
]

for col in categorical_missing_columns:
    if col in df.columns:
        mode_value = df[col].mode()[0]
        df[col] = df[col].fillna(mode_value)
        print(f"Filled missing values in {col} with mode: {mode_value}")
    else:
        print(f"Column not found: {col}")
```

```python
# Cell 9 - Confirm Missing Values Fixed

df.isnull().sum().sort_values(ascending=False)
```

```python
# Cell 10 - Check Duplicate Rows

duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# Cell 11 - Remove Duplicates If Found

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")

print("Dataset shape after duplicate check:", df.shape)
```

```python
# Cell 12 - Numerical Summary

df.describe().T
```

```python
# Cell 13 - Check Invalid Exam Score Values

print("Minimum Exam Score:", df["Exam_Score"].min())
print("Maximum Exam Score:", df["Exam_Score"].max())

invalid_exam_scores = df[(df["Exam_Score"] < 0) | (df["Exam_Score"] > 100)]

print("Number of invalid Exam_Score values:", len(invalid_exam_scores))
invalid_exam_scores
```

```python
# Cell 14 - Correct Invalid Exam Score

df["Exam_Score"] = df["Exam_Score"].clip(lower=0, upper=100)

print("Minimum Exam Score after correction:", df["Exam_Score"].min())
print("Maximum Exam Score after correction:", df["Exam_Score"].max())
```

```python
# Cell 15 - Create Target Variable

df["Performance_Label"] = np.where(df["Exam_Score"] >= 60, "Pass", "Fail")

df["Pass_Fail"] = np.where(df["Exam_Score"] >= 60, 1, 0)

df[["Exam_Score", "Performance_Label", "Pass_Fail"]].head(10)
```

```python
# Cell 16 - Check Target Distribution

target_counts = df["Performance_Label"].value_counts()
target_percentage = df["Performance_Label"].value_counts(normalize=True) * 100

target_distribution = pd.DataFrame({
    "Count": target_counts,
    "Percentage (%)": target_percentage.round(2)
})

target_distribution
```

```python
# Cell 17 - Final Validation Checks

required_columns = [
    "Exam_Score",
    "Performance_Label",
    "Pass_Fail"
]

print("Final Validation Checks")
print("-" * 50)

for col in required_columns:
    assert col in df.columns, f"Missing required column: {col}"
    print(f"{col}: Found")

missing_total = df.isnull().sum().sum()
assert missing_total == 0, "There are still missing values in the dataset."
print("Missing values: 0")

duplicate_total = df.duplicated().sum()
assert duplicate_total == 0, "There are duplicate rows in the dataset."
print("Duplicate rows: 0")

assert df["Exam_Score"].between(0, 100).all(), "Exam_Score contains invalid values."
print("Exam_Score range: Valid between 0 and 100")

assert set(df["Performance_Label"].unique()).issubset({"Pass", "Fail"}), "Invalid labels found."
assert set(df["Pass_Fail"].unique()).issubset({0, 1}), "Invalid numeric target values found."
print("Target labels: Valid")

print("\nAll final validation checks passed successfully.")
```

```python
# Cell 18 - Check Final Dataset

print("Final Dataset Shape:")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

df.head()
```

```python
# Cell 19 - Save Cleaned Dataset

PROCESSED_DATA_PATH.parent.mkdir(parents=True, exist_ok=True)

df.to_csv(PROCESSED_DATA_PATH, index=False)

print("Cleaned dataset saved successfully.")
print("Saved to:", PROCESSED_DATA_PATH)
```

```python
# Cell 20 - Save Data Cleaning Summary

cleaning_summary_path = TABLES_DIR / "data_cleaning_summary.txt"

cleaning_summary = f"""
Data Cleaning Summary

Dataset: StudentPerformanceFactors.csv

Final dataset shape:
Rows: {df.shape[0]}
Columns: {df.shape[1]}

Cleaning steps completed:
1. Loaded the raw dataset.
2. Checked dataset structure and data types.
3. Identified and handled missing categorical values.
4. Checked duplicate records.
5. Corrected invalid Exam_Score values using clipping between 0 and 100.
6. Created Performance_Label:
   - Pass if Exam_Score >= 60
   - Fail if Exam_Score < 60
7. Created Pass_Fail:
   - Pass = 1
   - Fail = 0
8. Saved the cleaned dataset.

Final checks:
Missing values: {df.isnull().sum().sum()}
Duplicate rows: {df.duplicated().sum()}
Minimum Exam_Score: {df["Exam_Score"].min()}
Maximum Exam_Score: {df["Exam_Score"].max()}

Target distribution:
{df["Performance_Label"].value_counts().to_string()}
"""

with open(cleaning_summary_path, "w", encoding="utf-8") as file:
    file.write(cleaning_summary)

print("Data cleaning summary saved successfully.")
print("Saved to:", cleaning_summary_path)
```

```python
# Cell 21 - Reload and Verify Saved File

df_cleaned = pd.read_csv(PROCESSED_DATA_PATH)

print("Reloaded cleaned dataset shape:", df_cleaned.shape)
df_cleaned.head()
```

```python
# Cell 22 - Final Cleaning Summary

print("DATA CLEANING SUMMARY")
print("-" * 40)
print(f"Final dataset rows: {df_cleaned.shape[0]}")
print(f"Final dataset columns: {df_cleaned.shape[1]}")
print("Missing values after cleaning:", df_cleaned.isnull().sum().sum())
print("Duplicate rows after cleaning:", df_cleaned.duplicated().sum())
print("Minimum Exam Score:", df_cleaned["Exam_Score"].min())
print("Maximum Exam Score:", df_cleaned["Exam_Score"].max())
print("\nTarget Distribution:")
print(df_cleaned["Performance_Label"].value_counts())
```
